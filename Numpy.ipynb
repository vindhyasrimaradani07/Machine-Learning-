```python
import matplotlib.pyplot as plt

plt.imshow(img_array)
plt.axis("off")
plt.show()
```

```python
height, width = img_array.shape[:2]

start_y = (height - 200) // 2
start_x = (width - 200) // 2

crop = img_array[start_y:start_y + 200, start_x:start_x + 200]

plt.imshow(crop)
plt.axis("off")
plt.show()
```

```python
vertical_flip = img_array[::-1]

plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")
plt.show()
```

```python
horizontal_flip = img_array[:, ::-1]

plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()
```

```python
red_channel = img_array[:, :, 0]

plt.imshow(red_channel, cmap="gray")
plt.title("Red Channel")
plt.axis("off")
plt.show()
```

```python
green_channel = img_array[:, :, 1]

plt.imshow(green_channel, cmap="gray")
plt.title("Green Channel")
plt.axis("off")
plt.show()
```

```python
blue_channel = img_array[:, :, 2]

plt.imshow(blue_channel, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")
plt.show()
```

```python
overall_brightness = np.mean(img_array)

print("Overall brightness:", overall_brightness)
```

```python
green_brightness = np.mean(green_channel)

print("Green brightness:", green_brightness)
```

```python
red_brightness = np.mean(red_channel)

print("Red brightness:", red_brightness)
```

```python
blue_brightness = np.mean(blue_channel)

print("Blue brightness:", blue_brightness)
```

```python
if red_brightness > green_brightness and red_brightness > blue_brightness:
    print("Brightest channel: Red")
elif green_brightness > red_brightness and green_brightness > blue_brightness:
    print("Brightest channel: Green")
else:
    print("Brightest channel: Blue")
```

```python
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"

tokens2 = sentence2.lower().split()

print(tokens2)
```

```python
tokens2_array = np.array(tokens2)

vocabulary2 = np.unique(tokens2_array)

print("Vocabulary:", vocabulary2)
print("Vocabulary size:", len(vocabulary2))
```

```python
tokens2_array = np.array(tokens2)

vec2 = np.array([
    np.sum(tokens2_array == word) for word in vocabulary2
])

print("Vocabulary:", vocabulary2)
print("Bag-of-words vector:", vec2)
```

```python
sentence1 = "I enjoy learning Python programming"
sentence2 = "I like studying Python coding"

tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()

combined_vocab = np.unique(np.array(tokens1 + tokens2))

vec1 = np.array([
    np.sum(np.array(tokens1) == word) for word in combined_vocab
])

vec2 = np.array([
    np.sum(np.array(tokens2) == word) for word in combined_vocab
])

cosine_sim = np.dot(vec1, vec2) / (
    np.linalg.norm(vec1) * np.linalg.norm(vec2)
)

print("Cosine similarity:", cosine_sim)
```

```python
paragraph = "Python is useful for data science. Python is popular for data analysis. Data science uses Python for analysis."

clean_text = paragraph.lower().replace(".", "")
tokens = clean_text.split()

tokens_array = np.array(tokens)

vocabulary = np.unique(tokens_array)

word_counts = np.array([
    np.sum(tokens_array == word) for word in vocabulary
])

sorted_indices = np.argsort(word_counts)
sorted_indices = sorted_indices[::-1]

top_3_indices = sorted_indices[:3]

for index in top_3_indices:
    print(vocabulary[index], ":", word_counts[index])
```
